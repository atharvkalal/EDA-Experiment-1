```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
sns.set_theme(style="whitegrid", palette="Set2")
plt.rcParams['figure.figsize'] = (9, 5)
plt.rcParams['font.size'] = 11

# 1. Load Dataset
df = pd.read_csv("house_price (1).csv")
print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head(5))

print("\nLast 5 Rows:")
display(df.tail(5))

# 2. Data Quality Check (Missing Values & Duplicates)
print("\nMissing Values:")
print(df.isnull().sum())

print("\nMissing Value Percentage:")
print((df.isnull().mean() * 100).round(2))

print("\nDuplicate Records:", df.duplicated().sum())

# 3. Descriptive Statistics
print("\nDescriptive Statistics:")
display(df.describe().round(2))

print("\nSkewness:")
display(df.skew(numeric_only=True))

# 4. Histograms of Numerical Features
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols].hist(
    bins=20,
    figsize=(16, 12),
    edgecolor='black',
    color='skyblue'
)
plt.suptitle("Distribution of Numerical Features (House Price Dataset)", fontsize=16)
plt.tight_layout()
plt.savefig("histograms.png", dpi=150)
plt.show()

# 5. Box Plots for Outlier Visualization
fig, axes = plt.subplots(4, 3, figsize=(15, 14))
axes = axes.flatten()
for idx, col in enumerate(numeric_cols):
    sns.boxplot(x=df[col], color='lightblue', ax=axes[idx])
    axes[idx].set_title(f"Box Plot of {col}")
axes[-1].axis('off')  # Hide unused 12th subplot
plt.tight_layout()
plt.savefig("boxplots.png", dpi=150)
plt.show()

# 6. Scatter Plot: Area vs. House Price (colored by Bedrooms)
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x='Area_sqft',
    y='Price_lakhs',
    hue='Bedrooms',
    palette='viridis',
    s=60,
    alpha=0.75
)
plt.title("Area (sq. ft.) vs House Price (Lakhs)")
plt.xlabel("Area (sq. ft.)")
plt.ylabel("Price (Lakhs)")
plt.tight_layout()
plt.savefig("area_vs_price.png", dpi=150)
plt.show()

# 7. Bar Plot: Average House Price by Number of Bedrooms
plt.figure(figsize=(9, 5))
sns.barplot(
    data=df,
    x='Bedrooms',
    y='Price_lakhs',
    errorbar=('ci', 95)
)
plt.title("Average House Price by Number of Bedrooms")
plt.xlabel("Number of Bedrooms")
plt.ylabel("Average Price (Lakhs)")
plt.tight_layout()
plt.savefig("price_by_bedrooms.png", dpi=150)
plt.show()

# 8. Bubble Chart: Area, Price, and Buyer Income
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x='Area_sqft',
    y='Price_lakhs',
    size='Income_lakhs',
    hue='Garage_cars',
    sizes=(30, 250),
    alpha=0.6
)
plt.title("Bubble Chart: Area, House Price, Income & Garage Capacity")
plt.xlabel("Area (sq. ft.)")
plt.ylabel("Price (Lakhs)")
plt.tight_layout()
plt.savefig("bubble_chart.png", dpi=150)
plt.show()

# 9. Correlation Heatmap
correlation = df.corr()
plt.figure(figsize=(11, 8))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1
)
plt.title("Correlation Heatmap - House Price Dataset")
plt.tight_layout()
plt.savefig("correlation_heatmap.png", dpi=150)
plt.show()

# 10. IQR Outlier Detection & Target Correlation Ranking
print("\n--- IQR OUTLIER DETECTION ---")
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col:<15} | Bounds: [{lower:7.2f}, {upper:7.2f}] | Outliers: {len(outliers)}")

score_corr = correlation['Price_lakhs'].drop('Price_lakhs').sort_values(ascending=False)
print("\nCorrelation with Price_lakhs:")
display(score_corr.to_frame("Correlation"))

# 11. Final Summary
print("=" * 50)
print("EXPLORATORY DATA ANALYSIS SUMMARY")
print("=" * 50)
print("Dataset Shape       :", df.shape)
print("Total Missing Values:", df.isnull().sum().sum())
print("Duplicate Records   :", df.duplicated().sum())
print("\nHouse Price Statistics (Lakhs):")
print(df['Price_lakhs'].describe().round(2))
```
